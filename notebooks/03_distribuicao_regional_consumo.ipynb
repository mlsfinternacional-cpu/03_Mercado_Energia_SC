## Base de análise

Com a integração entre a base de consumo da CELESC e a Divisão Territorial Brasileira (IBGE), foi possível incorporar à base de consumo as informações geográficas necessárias para a análise regional.

A base utilizada nesta etapa contempla 274 municípios com correspondência direta entre as duas fontes de dados, oferecendo ampla cobertura do território catarinense e garantindo consistência para as análises espaciais desenvolvidas a seguir.

A partir desta base consolidada, as próximas análises buscarão identificar como o consumo de energia elétrica está distribuído entre as diferentes regiões de Santa Catarina, destacando padrões de concentração, participação regional e possíveis diferenças na demanda energética do estado.

```python
import datetime
```

```python
# Seleciona automaticamente a última coluna temporal
coluna_data = max(col for col in base_sc.columns if isinstance(col, datetime.datetime))

consumo_regiao = (
    base_sc.groupby("Nome Região Geográfica Intermediária")[coluna_data]
    .sum()
    .reset_index()
    .rename(columns={coluna_data: "Consumo (MWh)"})
)

consumo_regiao["Participação (%)"] = (
    consumo_regiao["Consumo (MWh)"] / consumo_regiao["Consumo (MWh)"].sum() * 100
)

consumo_regiao = consumo_regiao.sort_values(
    by="Consumo (MWh)", ascending=False
).reset_index(drop=True)

consumo_regiao
```

```python
consumo_regiao["Consumo (MWh)"] = consumo_regiao["Consumo (MWh)"].round(2)
consumo_regiao["Participação (%)"] = consumo_regiao["Participação (%)"].round(2)

consumo_regiao
```

### Primeiras evidências

A distribuição regional do consumo de energia elétrica em Santa Catarina apresenta diferenças expressivas entre as Regiões Geográficas Intermediárias.

Observa-se uma forte concentração do consumo nas regiões de Blumenau e Joinville, que juntas representam mais da metade do consumo estadual no período analisado. Em seguida aparecem Chapecó, Florianópolis e Criciúma, enquanto Caçador e Lages apresentam participações significativamente menores.

Esses resultados indicam que a demanda energética do estado não está distribuída de forma homogênea, sugerindo a existência de polos regionais de maior concentração econômica e consumo de energia elétrica.

### Distribuição regional do consumo de energia elétrica

A tabela a seguir apresenta o consumo total de energia elétrica por Região Geográfica Intermediária de Santa Catarina, considerando o período mais recente disponível na base de dados.

Além do consumo absoluto, foi calculada a participação percentual de cada região em relação ao consumo estadual, permitindo identificar a concentração regional da demanda energética.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    consumo_regiao["Nome Região Geográfica Intermediária"],
    consumo_regiao["Consumo (MWh)"],
)

plt.title("Consumo de energia elétrica por Região Geográfica Intermediária")
plt.xlabel("Consumo (MWh)")
plt.ylabel("")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.show()
```

### Principais evidências

A análise evidencia uma distribuição regional desigual do consumo de energia elétrica em Santa Catarina. As Regiões Geográficas Intermediárias de Blumenau e Joinville concentram os maiores volumes de consumo do estado.

Em conjunto, essas duas regiões representam **52,55% do consumo estadual de energia elétrica**, evidenciando uma expressiva concentração regional da demanda energética.

As demais regiões apresentam participações inferiores, indicando que o consumo está distribuído de forma heterogênea no território catarinense. Essa configuração sugere a existência de polos regionais de maior intensidade econômica e energética, que serão explorados nas etapas seguintes da análise.

```python
# Participação acumulada das duas maiores regiões

top2 = consumo_regiao.head(2)

top2["Participação (%)"].sum()
```

```python
consumo_imediata = (
    base_sc.groupby("Nome Região Geográfica Imediata")[coluna_data]
    .sum()
    .reset_index()
    .rename(columns={coluna_data: "Consumo (MWh)"})
)

consumo_imediata["Participação (%)"] = (
    consumo_imediata["Consumo (MWh)"] / consumo_imediata["Consumo (MWh)"].sum() * 100
)

consumo_imediata = consumo_imediata.sort_values(
    by="Consumo (MWh)", ascending=False
).reset_index(drop=True)

consumo_imediata["Consumo (MWh)"] = consumo_imediata["Consumo (MWh)"].round(2)
consumo_imediata["Participação (%)"] = consumo_imediata["Participação (%)"].round(2)

consumo_imediata
```

### Distribuição do consumo por Região Geográfica Imediata

A análise por Região Geográfica Imediata permite aprofundar a compreensão da distribuição espacial do consumo de energia elétrica em Santa Catarina. Enquanto as Regiões Geográficas Intermediárias oferecem uma visão mais agregada do estado, a divisão imediata evidencia os principais polos regionais de demanda energética.

### Principais evidências

A análise por Região Geográfica Imediata reforça a concentração espacial do consumo de energia elétrica em Santa Catarina.

As cinco regiões imediatas com maior demanda — Joinville, Florianópolis, Itajaí, Blumenau e Criciúma — concentram **59,73%** do consumo estadual de energia elétrica.

Esse resultado evidencia que a demanda energética catarinense está fortemente concentrada em um número reduzido de polos regionais, indicando a existência de áreas com elevada intensidade econômica e energética.

```python
import matplotlib.pyplot as plt

top10 = consumo_imediata.head(10)

plt.figure(figsize=(10, 6))

plt.barh(top10["Nome Região Geográfica Imediata"], top10["Consumo (MWh)"])

plt.title("Consumo de energia elétrica por Região Geográfica Imediata")
plt.xlabel("Consumo (MWh)")
plt.ylabel("")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

### Principais evidências

O gráfico evidencia que o consumo de energia elétrica permanece concentrado em poucas Regiões Geográficas Imediatas. Joinville, Florianópolis e Itajaí destacam-se como os principais polos regionais de demanda, seguidas por Blumenau e Criciúma.

A partir da sexta posição observa-se uma redução gradual do consumo, reforçando a concentração da demanda energética em um conjunto restrito de regiões do estado.

## Conclusão

A análise regional do consumo de energia elétrica evidenciou uma distribuição espacial heterogênea da demanda energética em Santa Catarina.

Nas Regiões Geográficas Intermediárias, observou-se que Blumenau e Joinville concentram **52,55%** do consumo estadual. Ao aprofundar a análise para as Regiões Geográficas Imediatas, verificou-se que apenas cinco regiões — Joinville, Florianópolis, Itajaí, Blumenau e Criciúma — respondem por **59,73%** da demanda de energia elétrica do estado.

Os resultados indicam que o consumo estadual está fortemente concentrado em um conjunto reduzido de polos regionais, refletindo diferenças na distribuição das atividades econômicas, da urbanização e da demanda por energia entre as diversas regiões catarinenses.

Essas evidências ampliam a compreensão da organização territorial do consumo de energia elétrica em Santa Catarina e constituem uma base consistente para análises futuras envolvendo infraestrutura elétrica, economia regional, planejamento energético e identificação de áreas estratégicas para novos investimentos.
