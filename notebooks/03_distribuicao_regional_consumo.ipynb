---

## Objetivo

Investigar como o consumo de energia elétrica se distribui entre as diferentes regiões de Santa Catarina, buscando identificar concentrações regionais da demanda energética e possíveis diferenças entre as diversas áreas do estado.

Esta etapa amplia a análise espacial realizada anteriormente, permitindo compreender o consumo sob uma perspectiva regional, em vez da escala municipal.

---

## Motivação

Após compreender como o consumo está distribuído entre os municípios catarinenses e como essa demanda evoluiu ao longo do tempo, o próximo passo é analisar a organização regional do consumo de energia elétrica.

A análise regional permite reduzir a fragmentação observada na escala municipal e facilita a identificação de padrões territoriais associados ao desenvolvimento econômico, à infraestrutura energética e às diferentes características produtivas de Santa Catarina.

Além de aprofundar o entendimento sobre a demanda energética estadual, esta etapa fornecerá uma base importante para investigações futuras envolvendo economia regional, infraestrutura crítica, economia digital e potenciais áreas de interesse para implantação de Data Centers.

## Estratégia de Análise

Para responder a esta pergunta serão realizadas análises buscando identificar:

- participação relativa de cada região no consumo estadual;
- comparação entre regiões com maior e menor demanda energética;
